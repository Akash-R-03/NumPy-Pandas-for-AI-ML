# 1. Rename Columns

## Concept Explanation

Renaming columns means changing the existing column names of a DataFrame to names that are clearer or more suitable for the analysis.

Pandas provides the `rename()` method to rename one or more columns without changing the actual data.

Clear column names make a dataset easier to understand and work with.

## Real-world Example

Suppose Akash receives a dataset with column names like `Stud_Name` and `Stud_Marks`. He can rename them to `Name` and `Marks` to make the dataset easier to read.

## Business Example

A company may receive data with short or unclear column names. Renaming them can make reports and analysis easier for different teams to understand.

## AI/ML Example

In AI/ML projects, clear feature names make data preprocessing and model development easier to understand and maintain.

## Python Implementation

The example below renames two columns in a DataFrame.

## Output Explanation

The column names are changed while the actual data remains the same.

## Best Practices

- Use clear and meaningful column names.
- Keep a consistent naming style.
- Rename columns before starting complex analysis when the original names are unclear.

## Common Mistakes

- Misspelling the existing column name.
- Accidentally renaming the wrong column.
- Changing column names without updating code that depends on the old names.

## Advantages

- Improves dataset readability.
- Makes code easier to understand.
- Helps maintain consistent naming.

## Limitations

- Renaming does not change the actual data.
- Existing code using the old column names may need to be updated.

```python
import pandas as pd

df = pd.DataFrame({
    "Stud_Name": ["Akash R", "Rahul"],
    "Stud_Marks": [85, 78]
})

df = df.rename(columns={
    "Stud_Name": "Name",
    "Stud_Marks": "Marks"
})

print(df)
```

# 2. Drop Rows

## Concept Explanation

Dropping rows means removing selected rows from a DataFrame when they are not needed.

Pandas provides the `drop()` method to remove rows using their index labels.

This is useful when we want to remove unwanted, incorrect, or unnecessary records.

## Real-world Example

Suppose Akash has a student dataset containing one record that is no longer required. He can remove that row using its index.

## Business Example

A company may remove cancelled transactions or records that are no longer relevant to a particular analysis.

## AI/ML Example

In AI/ML preprocessing, unwanted or invalid records may need to be removed before the data is used for further analysis or model training.

## Python Implementation

The example below removes one row from a DataFrame.

## Output Explanation

The selected row is removed, while the remaining records stay unchanged.

## Best Practices

- Check the row before removing it.
- Keep a copy of the original data when the original records may be needed later.
- Use index labels carefully.

## Common Mistakes

- Dropping the wrong row because of an incorrect index.
- Removing data without checking whether it is actually unnecessary.
- Forgetting that `drop()` returns a new DataFrame unless `inplace=True` is used.

## Advantages

- Simple way to remove unwanted records.
- Useful during data cleaning.
- Helps create a focused dataset.

## Limitations

- Incorrect row selection can result in data loss.
- Removing rows does not automatically solve the reason why they were unwanted.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "Marks": [85, 78, 92]
})

df = df.drop(index=1)

print(df)
```

# 3. Drop Columns

## Concept Explanation

Dropping columns means removing one or more columns from a DataFrame when they are not needed.

Pandas provides the `drop()` method with `axis=1` to remove columns.

Removing unnecessary columns can make a dataset simpler and easier to work with.

## Real-world Example

Suppose Akash has a student dataset containing names, marks, and an extra column that is not required. He can remove that column before continuing the analysis.

## Business Example

A company may receive a dataset containing many fields that are not relevant to a particular report. Unnecessary columns can be removed to focus on the required information.

## AI/ML Example

In AI/ML, unnecessary columns can be removed during data preparation when they do not provide useful information for the analysis or model.

## Python Implementation

The example below removes the `Age` column from a DataFrame.

## Output Explanation

The selected column is removed, while the remaining columns stay unchanged.

## Best Practices

- Check whether a column is actually unnecessary before removing it.
- Keep the original dataset when the removed information may be needed later.
- Remove irrelevant columns before further processing when appropriate.

## Common Mistakes

# 4. Insert Columns

Inserting a column means adding a new column at a specific position in a Pandas DataFrame.

# 5. assign()

## Concept Explanation

The `assign()` method is used to create new columns or modify existing columns in a Pandas DataFrame.

It returns a new DataFrame instead of changing the original DataFrame directly.

This is useful when we want to create calculated columns while keeping the original data unchanged.

## Real-world Example

Suppose Akash has student marks and wants to create a new column showing whether each student passed or failed.

## Business Example

A company can use `assign()` to create calculated fields such as profit, discount, or customer category from existing data.

## AI/ML Example

In AI/ML, `assign()` can be useful for creating new features from existing columns during data preprocessing.

## Python Implementation

The example below creates a new `Result` column based on the marks.

## Output Explanation

A new column is added to the DataFrame based on the existing `Marks` values.

## Best Practices

- Use meaningful names for newly created columns.
- Keep calculations simple and readable.
- Remember that `assign()` returns a new DataFrame.

## Common Mistakes

- Assuming the original DataFrame is automatically changed.
- Using incorrect column names in calculations.
- Creating features without checking whether they are meaningful.

## Advantages

- Simple way to create new columns.
- Useful for feature creation.
- Supports readable data-processing workflows.

## Limitations

- Can become harder to read when many complex calculations are placed together.
- The original DataFrame remains unchanged unless the result is assigned back.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

result = df.assign(Result=df["Marks"] >= 40)

print(result)
```

# 6. apply()

## Concept Explanation

The `apply()` method is used to apply a function to values in a Pandas Series or DataFrame.

It is useful when we want to perform the same custom operation on multiple values.

## Real-world Example

Suppose Akash has student marks and wants to convert each mark into a simple grade category. `apply()` can apply the same grading logic to every mark.

## Business Example

A company can use `apply()` to categorize customers based on their purchase amount or assign a status based on business rules.

## AI/ML Example

In AI/ML, `apply()` can be useful for transforming or creating features when a custom operation is needed during data preprocessing.

## Python Implementation

The example below creates a `Grade` column based on the marks.

## Output Explanation

The function is applied to each value in the `Marks` column and the resulting grades are stored in a new column.

## Best Practices

- Keep the function simple and readable.
- Use vectorized Pandas operations when they can perform the same task efficiently.
- Make sure the function handles the expected data correctly.

## Common Mistakes

- Using `apply()` for operations that Pandas already provides directly.
- Writing overly complex functions.
- Applying a function to the wrong column or axis.

## Advantages

- Allows custom operations.
- Easy to understand for simple transformations.
- Useful when built-in methods are not enough.

## Limitations

- Can be slower than vectorized operations for large datasets.
- Complex functions can make the code harder to maintain.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "Marks": [85, 62, 92]
})

df["Grade"] = df["Marks"].apply(lambda x: "A" if x >= 80 else "B")

print(df)
```

# 7. map()

## Concept Explanation

The `map()` method is used to transform values in a Pandas Series by applying a function, dictionary, or mapping rule to each value.

It is commonly used when we want to replace or convert individual values in a single column.

## Real-world Example

Suppose Akash has student scores represented by simple categories such as `A`, `B`, and `C`. He can use `map()` to convert those categories into meaningful descriptions.

## Business Example

A company can use `map()` to convert short codes such as `P`, `C`, and `R` into descriptions like Pending, Completed, and Rejected.

## AI/ML Example

In AI/ML, `map()` can be used to convert categorical values into suitable representations before further preprocessing.

## Python Implementation

The example below maps grade codes to their full descriptions.

## Output Explanation

Each grade code is replaced with its corresponding description.

## Best Practices

- Use clear mapping rules.
- Check for values that are not included in the mapping.
- Use `map()` mainly for Series-level transformations.

## Common Mistakes

- Missing values in the mapping dictionary.
- Using `map()` when a DataFrame-level operation is required.
- Assuming unmapped values will automatically receive a meaningful value.

## Advantages

- Simple way to transform individual column values.
- Easy to read for small mapping rules.
- Useful for categorical data.

## Limitations

- Mainly works on a Series.
- Unmapped dictionary values can become missing values.
- More complex transformations may require other methods.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "Grade": ["A", "B", "A"]
})

grade_map = {"A": "Excellent", "B": "Good"}

df["Performance"] = df["Grade"].map(grade_map)

print(df)
```

# 8. replace()

## Concept Explanation

The `replace()` method is used to replace existing values in a Pandas DataFrame or Series with new values.

It is useful when we need to correct values, standardize data, or change specific values in a dataset.

## Real-world Example

Suppose Akash has a student dataset where a city is written as `B'lore`. He can use `replace()` to change it to the more consistent value `Bangalore`.

## Business Example

A company may receive customer data with different versions of the same category, such as `Completed` and `complete`. These values can be standardized using `replace()`.

## AI/ML Example

In AI/ML preprocessing, `replace()` can be used to correct inconsistent categorical values before the data is used for analysis or model training.

## Python Implementation

The example below replaces `B'lore` with `Bangalore`.

## Output Explanation

The incorrect or inconsistent value is replaced with the required value.

## Best Practices

- Check the existing values before replacing them.
- Use consistent replacement values.
- Verify the result after replacement.

## Common Mistakes

- Replacing the wrong value.
- Using inconsistent replacement values.
- Changing valid values accidentally.

## Advantages

- Simple way to correct values.
- Useful for standardizing categorical data.
- Can replace specific values across a DataFrame.

## Limitations

- Incorrect replacement rules can change valid data.
- It does not automatically identify all inconsistent values.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "City": ["B'lore", "Chennai", "B'lore"]
})

df["City"] = df["City"].replace("B'lore", "Bangalore")

print(df)
```

# 9. Duplicate Handling

## Concept Explanation

Duplicate handling means finding and managing rows that contain the same information more than once.

Pandas provides methods such as `duplicated()` to identify duplicate rows and `drop_duplicates()` to remove them.

Removing unnecessary duplicates helps keep the dataset clean and avoids counting the same record multiple times.

## Real-world Example

Suppose Akash has a student dataset where the same student record was entered twice. Duplicate handling can be used to identify and remove the repeated record.

## Business Example

A company may receive duplicate customer or transaction records from different sources. Removing duplicates helps maintain accurate reports and analysis.

## AI/ML Example

In AI/ML, duplicate records can cause certain observations to appear more frequently than they should. Handling duplicates can help create a cleaner dataset for analysis and model training.

## Python Implementation

The example below identifies and removes duplicate student records.

## Output Explanation

The duplicate record is removed, leaving only one copy of each record.

## Best Practices

- Check duplicates before removing them.
- Understand why duplicates exist before deleting them.
- Keep the original dataset when the removed records may be needed later.

## Common Mistakes

- Removing valid repeated records by mistake.
- Checking only one column when the complete row should be compared.
- Deleting duplicates without first understanding the dataset.

## Advantages

- Helps keep data clean.
- Prevents duplicate records from affecting analysis.
- Useful during data preprocessing.

## Limitations

- Not every repeated value is a duplicate record.
- Incorrect duplicate rules can remove useful information.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Akash R"],
    "Marks": [85, 78, 85]
})

print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print(df)
```
