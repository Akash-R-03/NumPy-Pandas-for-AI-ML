# 2. Random Seed

## Concept Explanation

A random seed is a fixed value used to control the starting point of random number generation.

Normally, random functions can produce different values each time we run the code. When we set the same seed, NumPy produces the same sequence of random values each time.

This is useful when we want our results to be reproducible.

## Real-world Example

Suppose we are running a simulation several times and want to compare the results. Using the same seed allows us to generate the same random values each time.

## Business Example

A company may use random sampling while testing a business process. Setting a seed makes the sample consistent so that different team members can reproduce the same analysis.

## AI/ML Example

Random seeds are commonly used in Machine Learning experiments. They help make operations involving randomness reproducible, which makes it easier to compare models and debug results.

## Python Implementation

## Output Explanation

Because the seed is fixed, the same random values will be generated every time the code is run.

## Best Practices

- Set a seed when reproducibility is important.
- Use the same seed when you need to repeat an experiment.
- Clearly document the seed used in an experiment.

## Common Mistakes

- Expecting random results to remain the same without setting a seed.
- Changing the seed when trying to reproduce an earlier result.

## Advantages

- Makes experiments reproducible.
- Helps with debugging.
- Makes comparisons between experiments easier.

## Limitations

- Using a fixed seed does not make the numbers truly random.
- Different libraries or random-generation methods may require their own seed settings.

```python
import numpy as np

np.random.seed(42)

numbers = np.random.rand(5)

print("Random numbers:", numbers)
```

# 4. Random Choice

## Concept Explanation

Random choice is used to randomly select one or more values from an array or a given set of values.

NumPy provides the `np.random.choice()` function for this.

We can use it when we want to randomly select data from a group of available options.

## Real-world Example

Suppose we have a list of fruits and want to randomly select one fruit. Random choice can do this for us.

## Business Example

A company can randomly select customers from a customer list for a survey or promotional activity.

## AI/ML Example

Random choice can be used to select random samples from a dataset for testing, analysis, or experimentation.

## Output Explanation

The output contains randomly selected values from the original array.

## Best Practices

- Use a seed when reproducible results are required.
- Make sure the values being selected are suitable for the task.
- Use `replace=False` when the same value should not be selected more than once.

## Common Mistakes

- Forgetting that sampling with replacement can select the same value more than once.
- Trying to select more unique values than are available when using `replace=False`.

## Advantages

- Makes random sampling simple.
- Useful for simulations and testing.
- Can select one or multiple values.

## Limitations

- Results can change between runs without a fixed seed.

```python
import numpy as np

np.random.seed(42)

items = np.array(["Apple", "Banana", "Mango", "Orange"])

selected = np.random.choice(items, 2, replace=False)

print("Selected items:", selected)
```

# 5. Shuffle

## Concept Explanation

Shuffling means randomly changing the order of the elements in an array.

NumPy provides `np.random.shuffle()` to shuffle an array in place. This means the original array itself is changed.

## Real-world Example

If we have a list of students and want to randomly arrange their order, we can shuffle the list.

## Business Example

A company can shuffle a customer list before randomly selecting customers for a survey or test.

## AI/ML Example

Shuffling data is commonly used before training Machine Learning models so that the model does not learn from data in a fixed order.

## Python Implementation

The code below randomly changes the order of the values in an array.

## Output Explanation

The elements remain the same, but their order changes.

## Best Practices

- Use a random seed when reproducible results are needed.
- Remember that `shuffle()` changes the original array.
- Keep a copy if the original order is important.

## Common Mistakes

- Assuming `shuffle()` creates a new array.
- Forgetting that the original array is modified.

## Advantages

- Simple way to randomize data order.
- Useful for sampling and Machine Learning workflows.

## Limitations

- The original array is changed.
- The result is not reproducible without a fixed seed.

```python
import numpy as np

np.random.seed(42)

data = np.array([1, 2, 3, 4, 5])

np.random.shuffle(data)

print("Shuffled data:", data)
```

# 6. Permutation

## Concept Explanation

Permutation is used to get a random arrangement of the elements in an array.

NumPy provides `np.random.permutation()` for this. Unlike `shuffle()`, it returns a new array and keeps the original array unchanged.

## Real-world Example

Suppose we have a list of students and want to create a random order for a presentation. A permutation can generate that order.

## Business Example

A company can randomly arrange customers or tasks before assigning them for testing or analysis.

## AI/ML Example

Permutation can be useful when randomly rearranging data while keeping the original data unchanged. It can also be useful in experiments where different feature or data orders need to be tested.

## Python Implementation

## Output Explanation

The new array contains the same values in a different order, while the original array remains unchanged.

## Best Practices

- Use a seed when reproducible results are required.
- Use permutation when you want a new randomized array.
- Use shuffle when changing the existing array is acceptable.

## Common Mistakes

- Confusing `permutation()` with `shuffle()`.
- Forgetting that `permutation()` returns a new array.

## Advantages

- Keeps the original array unchanged.
- Easy way to create a random ordering.
- Useful for experiments and data preparation.

## Limitations

- Results change between runs without a fixed seed.
- Random ordering does not guarantee a meaningful or balanced sample.

```python
import numpy as np

np.random.seed(42)

data = np.array([1, 2, 3, 4, 5])
random_order = np.random.permutation(data)

print("Original:", data)
print("Permutation:", random_order)
```

# 7. Random Distributions

## Concept Explanation

A random distribution describes how randomly generated values are spread.

NumPy provides different random distributions depending on the type of data we want to simulate.

For example, the normal distribution is useful when values tend to be concentrated around an average, while the uniform distribution gives values with equal likelihood across a specified range.

## Real-world Example

Heights of people in a large population can often be modelled approximately using a normal distribution.

## Business Example

A business can use random distributions to simulate customer demand, sales values, waiting times, or other situations where the exact outcome is uncertain.

## AI/ML Example

Random distributions are useful in AI/ML for generating sample data, testing algorithms, simulations, and understanding how data may be distributed.

## Python Implementation

The example below generates random values from a normal distribution.

## Output Explanation

The generated values are random and are generally centred around the specified mean.

## Best Practices

- Choose a distribution that matches the situation being modelled.
- Use a random seed when reproducibility is required.
- Understand the parameters of the selected distribution.

## Common Mistakes

- Assuming every dataset follows a normal distribution.
- Choosing a distribution without understanding the problem.
- Forgetting that random results can change between runs.

## Advantages

- Useful for simulations and testing.
- Helps generate realistic sample data.
- Supports many common probability distributions.

## Limitations

- A randomly generated dataset may not perfectly represent real-world data.
- Choosing the wrong distribution can produce misleading results.

```python
import numpy as np

np.random.seed(42)

values = np.random.normal(loc=50, scale=10, size=5)

print("Random values:", values)
```
