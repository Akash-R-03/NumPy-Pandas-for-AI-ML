# 1. CSV Files

## Concept Explanation

CSV stands for Comma-Separated Values. It is a simple file format commonly used to store tabular data in rows and columns.

CSV files are very common because they are easy to create, read, share, and open using different tools.

Pandas provides `read_csv()` to read CSV files and `to_csv()` to save DataFrames as CSV files.

## Real-world Example

Suppose Akash has a list of student records containing names, marks, and ages. This information can be stored in a CSV file and loaded into Pandas whenever it is needed.

## Business Example

Companies commonly store sales, customer, and transaction records in CSV files. These files can be shared between teams and loaded into Python for analysis.

## AI/ML Example

Many Machine Learning datasets are available in CSV format. Pandas can read the file into a DataFrame so that the data can be explored, cleaned, and prepared for a model.

## Python Implementation

The example below creates a small DataFrame, saves it as a CSV file, and reads the file back using Pandas.

## Output Explanation

The output shows the data after it has been read from the CSV file.

## Best Practices

- Use clear column names.
- Check the data after reading the file.
- Use `index=False` when saving a DataFrame if the index is not required.

## Common Mistakes

- Using an incorrect file path.
- Forgetting to check whether the CSV contains the expected columns.
- Accidentally saving the DataFrame index as an extra column.

## Advantages

- Simple and widely supported.
- Easy to share between applications.
- Works well with Pandas.

## Limitations

- CSV does not preserve all DataFrame information such as data types and indexes.
- Large CSV files can require significant memory when loaded completely.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df.to_csv("students.csv", index=False)

data = pd.read_csv("students.csv")

print(data)
```

# 3. JSON Files

JSON stands for JavaScript Object Notation. It is a lightweight format commonly used to store and exchange structured data.

# 5. Pickle Files

## Concept Explanation

Pickle is a Python format used to save Python objects and load them back later.

Unlike CSV or JSON, Pickle can preserve Python objects such as DataFrames, lists, dictionaries, and other Python structures.

Pandas provides `to_pickle()` to save a DataFrame and `read_pickle()` to load it again.

## Real-world Example

Suppose Akash has prepared a DataFrame and wants to save it without losing its Python structure. Pickle can store the DataFrame and load it again later.

## Business Example

A company can use Pickle to temporarily store processed Python data so that the same data can be reused without repeating the processing steps.

## AI/ML Example

Pickle is commonly used to save trained Machine Learning models or preprocessing objects so they can be loaded and used later.

## Python Implementation

The example below saves a DataFrame as a Pickle file and reads it back.

## Output Explanation

The output shows the DataFrame after it has been loaded from the Pickle file.

## Best Practices

- Use Pickle mainly within trusted Python environments.
- Keep track of the Python and library versions used.
- Use clear file names for saved objects.

## Common Mistakes

- Loading Pickle files from unknown or untrusted sources.
- Assuming Pickle files can be easily used by non-Python applications.
- Forgetting which Python object was stored in the file.

## Advantages

- Preserves Python objects easily.
- Simple to save and load DataFrames.
- Useful for Machine Learning workflows.

## Limitations

- Pickle is Python-specific.
- It is not designed for safe loading of untrusted files.
- It is less suitable for sharing data between different programming languages.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df.to_pickle("students.pkl")

data = pd.read_pickle("students.pkl")

print(data)
```

# 6. SQL Tables

## Concept Explanation

Pandas can work with data stored in SQL databases.

Using a database connection, we can read SQL query results directly into a Pandas DataFrame.

This is useful when the data is stored in a database instead of a CSV, Excel, or other file.

## Real-world Example

Suppose Akash's project data is stored in a SQL database. Pandas can retrieve the required records and convert them into a DataFrame for analysis.

## Business Example

A company may store customer, sales, and transaction data in a database. Analysts can use SQL to select the required data and Pandas to perform further analysis.

## AI/ML Example

In AI/ML projects, training data is often stored in databases. Pandas can retrieve the required records and prepare them for data cleaning and model development.

## Python Implementation

The example below shows the basic approach for reading SQL query results into Pandas.

## Output Explanation

The query result is loaded into a Pandas DataFrame.

## Best Practices

- Retrieve only the columns and rows needed.
- Use SQL filters to reduce unnecessary data transfer.
- Keep database credentials secure.

## Common Mistakes

- Loading an entire large table when only a small portion is needed.
- Hardcoding database passwords in notebooks.
- Using an incorrect SQL query or database connection.

## Advantages

- Directly works with database data.
- Useful for large structured datasets.
- Combines SQL querying with Pandas analysis.

## Limitations

- Requires a database connection and suitable database driver.
- Database access may require authentication and permissions.
- Large queries can take time and memory to load into Pandas.

```python
import pandas as pd
import sqlite3

connection = sqlite3.connect("students.db")

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df.to_sql("students", connection, index=False, if_exists="replace")

data = pd.read_sql("SELECT * FROM students", connection)

print(data)

connection.close()
```

# 7. Reading Multiple Files

## Concept Explanation

Sometimes data is stored in multiple files instead of a single file.

Pandas can be combined with Python's file-handling tools to find and read multiple files and combine their data into one DataFrame.

This is useful when data is divided into separate files, such as monthly sales files or daily reports.

## Real-world Example

Suppose Akash has separate CSV files for January, February, and March. Instead of reading each file manually, we can read all the files and combine them into one dataset.

## Business Example

A company may receive separate sales reports every month. These files can be combined to create a single dataset for analysis.

## AI/ML Example

An AI/ML project may have training data stored across multiple files. Combining the files can create a single dataset for data exploration and preprocessing.

## Python Implementation

The example below creates two CSV files and combines them into one DataFrame.

## Output Explanation

The output contains the records from both CSV files in a single DataFrame.

## Best Practices

- Make sure all files have a consistent structure.
- Check column names before combining files.
- Handle missing or invalid files carefully.

## Common Mistakes

- Combining files with different column names without checking them.
- Reading unrelated files from the same folder.
- Forgetting to verify the final combined dataset.

## Advantages

- Saves time when working with many files.
- Makes batch data processing easier.
- Useful for preparing larger datasets.

## Limitations

- Different file structures may require additional processing.
- Reading many large files can consume significant memory.

```python
import pandas as pd

file1 = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

file2 = pd.DataFrame({
    "Name": ["Kiran", "Anu"],
    "Marks": [92, 88]
})

file1.to_csv("students_1.csv", index=False)
file2.to_csv("students_2.csv", index=False)

data = pd.concat([
    pd.read_csv("students_1.csv"),
    pd.read_csv("students_2.csv")
], ignore_index=True)

print(data)
```
