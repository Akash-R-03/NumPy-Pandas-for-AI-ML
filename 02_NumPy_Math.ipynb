Mean is the average value of a group of numbers.

# 4. Variance

## Concept Explanation

Variance tells us how much the values in a dataset differ from their mean.

A small variance means the values are close to the average. A large variance means the values are more spread out.

NumPy provides the `np.var()` function to calculate variance.

## Real-world Example

If the daily temperatures are very close to each other, the variance will be small. If the temperatures change a lot from day to day, the variance will be larger.

## Business Example

A company can use variance to understand how much sales values change from one period to another.

## AI/ML Example

Variance helps us understand how much a feature varies in a dataset. It is also useful when analysing features during Machine Learning preprocessing.

## Python Implementation

## Output Explanation

The output represents the average squared difference of the values from their mean.

## Best Practice

Use variance when you need to understand how spread out numerical data is.

## Common Mistake

Variance is measured in squared units, so it is not always as easy to interpret as standard deviation.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

variance = np.var(data)

print("Variance:", variance)
```

# 5. Standard Deviation

## Concept Explanation

Standard deviation tells us how much the values in a dataset normally vary from the mean.

A small standard deviation means the values are close to the average. A large standard deviation means the values are more spread out.

Standard deviation is easier to interpret than variance because it is in the same units as the original data.

NumPy provides the `np.std()` function to calculate it.

## Real-world Example

If the daily temperatures are close to the average temperature, the standard deviation will be small. If the temperatures change significantly, it will be larger.

## Business Example

A company can use standard deviation to understand how much sales vary over different months.

## AI/ML Example

Standard deviation is commonly used to understand the spread of Machine Learning features and is also important in data preprocessing and standardization.

## Python Implementation

The code below calculates the standard deviation of a small dataset.

## Output Explanation

The output shows how much the values typically vary around their mean.

## Best Practice

Use standard deviation when you want an easy-to-understand measure of data spread.

## Common Mistake

Do not confuse standard deviation with variance. Standard deviation is the square root of variance.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

std = np.std(data)

print("Standard Deviation:", std)
```

# 10. Percentile

## Concept Explanation

A percentile tells us the value below which a certain percentage of the data falls.

For example, the 50th percentile represents the middle of the data and is related to the median.

NumPy provides the `np.percentile()` function to calculate percentiles.

## Real-world Example

If a student's score is at the 90th percentile, it means the score is higher than or equal to the scores of most students in the group.

## Business Example

A company can use percentiles to understand customer spending. For example, the 90th percentile can show a spending level that only a small percentage of customers exceed.

## AI/ML Example

# 11. Dot Product

## Concept Explanation

The dot product combines corresponding values from two arrays, multiplies them, and then adds the results together.

For example, for `[1, 2, 3]` and `[4, 5, 6]`:

`(1 × 4) + (2 × 5) + (3 × 6) = 32`

NumPy provides `np.dot()` to calculate the dot product.

## Real-world Example

Suppose we have the quantity of three products and their prices. The dot product can help calculate the total value of all the products.

## Business Example

## AI/ML Example

## Python Implementation

The code below calculates the dot product of two simple arrays.

## Output Explanation

# 13. Transpose

## Concept Explanation

Transpose means changing the rows of a matrix into columns and the columns into rows.

In NumPy, we can use `.T` to find the transpose of an array.

For example, a matrix with 2 rows and 3 columns becomes a matrix with 3 rows and 2 columns after transposing.

## Real-world Example

Suppose information is arranged with students as rows and subjects as columns. Transposing the data can change the arrangement so that subjects become rows.

## Business Example

A company may have sales data arranged by products and months. Transposing can change the structure so that months become rows and products become columns.

## AI/ML Example

Transpose operations are commonly used in Machine Learning and numerical computing when changing the orientation of matrices for calculations.

## Python Implementation

The code below finds the transpose of a simple matrix.

## Output Explanation

The rows of the original matrix become columns in the transposed matrix.

## Best Practice

Check the shape of the matrix before and after transposing when the structure matters.

## Common Mistake

Do not assume that transposing changes the actual values. It changes their arrangement.

```python
import numpy as np

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print("Original:\n", matrix)
print("Transpose:\n", matrix.T)
```

# Interview Questions

### 1. What is the difference between `np.mean()` and `np.median()`?

`np.mean()` calculates the average of the values, while `np.median()` returns the middle value after the data is arranged in order.

### 2. What is the difference between variance and standard deviation?

Variance measures the spread of data using squared differences from the mean. Standard deviation is the square root of variance and is easier to interpret because it uses the same units as the original data.

### 3. What is the difference between `np.min()` and `np.argmin()`?

`np.min()` returns the smallest value, while `np.argmin()` returns the index where the smallest value is located.

### 4. What is the difference between `np.max()` and `np.argmax()`?

`np.max()` returns the largest value, while `np.argmax()` returns the index where the largest value is located.

### 5. What is a dot product?

A dot product multiplies corresponding values from two arrays and adds the results together. It is commonly used in numerical computing and Machine Learning.

### 6. What is the difference between element-wise multiplication and matrix multiplication?

Element-wise multiplication multiplies corresponding elements using `*`. Matrix multiplication follows the row-column multiplication rule and can be performed using `@` or `np.matmul()`.

### 7. What does transpose do?

Transpose changes the rows of a matrix into columns and the columns into rows.

### 8. Why is linear algebra important in Machine Learning?

Linear algebra provides the mathematical operations needed to work with vectors and matrices. Operations such as dot products and matrix multiplication are used in many Machine Learning algorithms and neural networks.
