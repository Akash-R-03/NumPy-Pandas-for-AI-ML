# 1. loc

## Concept Explanation

The `loc` accessor is used to select rows and columns from a Pandas DataFrame using labels.

It is useful when we want to select data based on row labels or column names.

With `loc`, we can select a single row, multiple rows, specific columns, or filter data using conditions.

## Real-world Example

Suppose Akash has a student DataFrame and wants to select the record of a particular student. `loc` can be used to access the required row using its label.

## Business Example

A company can use `loc` to select specific customer records or particular columns from a business dataset.

## AI/ML Example

In AI/ML workflows, `loc` can be useful when selecting specific features or rows based on labels or conditions during data preparation.

## Python Implementation

The example below uses `loc` to select the row belonging to a particular index label.

## Output Explanation

The output shows the data stored in the selected row.

## Best Practices

- Use `loc` when working with row labels or column names.
- Use clear and meaningful indexes when label-based selection is important.
- Check column names before selecting them.

## Common Mistakes

- Confusing `loc` with `iloc`.
- Using a label that does not exist in the DataFrame.
- Forgetting that `loc` works with labels rather than integer positions.

## Advantages

- Easy to read and understand.
- Supports both row and column selection.
- Useful for label-based filtering.

## Limitations

- Requires correct row labels and column names.
- Can be confusing when the index contains integer labels.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "Marks": [85, 78, 92]
}, index=["S01", "S02", "S03"])

print(df.loc["S01"])
```

# 2. iloc

## Concept Explanation

The `iloc` accessor is used to select rows and columns from a Pandas DataFrame based on their integer positions.

The position starts from `0`, so the first row or column has position `0`.

Unlike `loc`, which works with labels, `iloc` works with the numerical position of the data.

## Real-world Example

Suppose Akash has a student DataFrame and wants to select the first student record. `iloc` can select it using its position.

## Business Example

A company can use `iloc` to select specific rows or columns based on their position when working with structured reports.

## AI/ML Example

In AI/ML data preparation, `iloc` can be useful when selecting a specific range of rows or columns based on their positions.

## Python Implementation

The example below uses `iloc` to select the first row of the DataFrame.

## Output Explanation

The output shows the first row because its integer position is `0`.

## Best Practices

- Use `iloc` when position-based selection is required.
- Remember that indexing starts from `0`.
- Use `loc` when working with labels instead.

## Common Mistakes

- Confusing `iloc` with `loc`.
- Forgetting that the first position is `0`.
- Trying to use column names directly with `iloc`.

## Advantages

- Simple for position-based selection.
- Useful when working with ranges of rows or columns.
- Works independently of the actual index labels.

## Limitations

- Requires knowledge of row or column positions.
- Positions can change when the DataFrame structure changes.

# 3. Boolean Indexing

## Concept Explanation

Boolean indexing is a way of selecting rows from a Pandas DataFrame using a condition.

The condition produces `True` or `False` values for each row. Pandas then returns only the rows where the condition is `True`.

It is one of the most commonly used techniques for filtering data.

## Real-world Example

Suppose Akash has a student dataset and wants to see only students who scored more than 80 marks. Boolean indexing can be used to filter those records.

## Business Example

A company can use Boolean indexing to find customers whose purchase amount is greater than a specific value.

## AI/ML Example

In AI/ML, Boolean indexing can be used to remove unwanted records, select data based on conditions, or create subsets for further analysis.

## Python Implementation

The example below selects students whose marks are greater than 80.

## Output Explanation

Only the rows satisfying the condition are returned.

## Best Practices

- Keep filtering conditions clear and readable.
- Use parentheses when combining multiple conditions.
- Check the filtered result before using it for further analysis.

## Common Mistakes

- Using `and` or `or` instead of `&` or `|` for Series conditions.
- Forgetting parentheses around individual conditions.
- Using the wrong comparison operator.

## Advantages

- Simple and powerful filtering method.
- Works well with numerical and categorical conditions.
- Useful for data analysis and preprocessing.

## Limitations

- Complex conditions can become difficult to read.
- Incorrect conditions may return unexpected results.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran"],
    "Marks": [85, 78, 92]
})

result = df[df["Marks"] > 80]

print(result)
```

# 4. Filtering Rows

## Concept Explanation

Filtering rows means selecting only the records that meet a particular condition.

In Pandas, we can filter rows using conditions on one or more columns. This helps us focus only on the data that is relevant to the current analysis.

## Real-world Example

Suppose Akash has a student dataset and wants to see only students who scored 80 or more marks. He can filter the rows using the `Marks` column.

## Business Example

A company can filter customer records to find customers from a particular city or customers whose purchase amount is above a certain value.

## AI/ML Example

In AI/ML, filtering rows can be used to remove unwanted records or select a specific group of observations before data preprocessing.

## Python Implementation

The example below filters students whose marks are 80 or above.

## Output Explanation

Only the rows that satisfy the condition are displayed.

## Best Practices

- Use clear conditions when filtering data.
- Check the filtered DataFrame before continuing the analysis.
- Use multiple conditions carefully when required.

## Common Mistakes

- Using the wrong comparison operator.
- Forgetting parentheses when combining conditions.
- Filtering the wrong column.

## Advantages

- Helps focus on relevant data.
- Simple and flexible.
- Useful in data analysis and preprocessing.

## Limitations

- Incorrect conditions can remove useful records.
- Complex filtering conditions can become harder to read.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "Marks": [85, 78, 92, 68]
})

filtered = df[df["Marks"] >= 80]

print(filtered)
```

# 5. query()

## Concept Explanation

The `query()` method is used to filter rows in a Pandas DataFrame using a condition written as a string.

It can make filtering easier to read, especially when the condition involves multiple columns.

## Real-world Example

Suppose Akash has student records and wants to find students who scored more than 80 marks and are below 25 years old.

## Business Example

A company can use `query()` to find customers who meet multiple conditions, such as having a high purchase amount and belonging to a particular location.

## AI/ML Example

In AI/ML, `query()` can be used to create a subset of data that meets specific conditions before further analysis or preprocessing.

## Python Implementation

The example below selects students who have marks above 80 and are below 25 years old.

## Output Explanation

Only the rows that satisfy both conditions are returned.

## Best Practices

- Keep query conditions simple and readable.
- Use `query()` when string-based filtering makes the condition easier to understand.
- Check the filtered result before using it.

## Common Mistakes

- Using incorrect column names.
- Writing the query condition incorrectly.
- Forgetting that column names with spaces may require special handling.

## Advantages

- Makes filtering conditions easy to read.
- Useful for multiple conditions.
- Reduces lengthy Boolean expressions.

## Limitations

- Query expressions have specific syntax rules.
- Very complex conditions may be clearer using normal Boolean indexing.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "Age": [24, 26, 23, 27],
    "Marks": [85, 78, 92, 88]
})

result = df.query("Marks > 80 and Age < 25")

print(result)
```

# 6. isin()

## Concept Explanation

The `isin()` method is used to check whether values in a Pandas column belong to a given list of values.

It is useful when we want to filter data based on multiple possible values without writing a separate condition for each value.

## Real-world Example

Suppose Akash has student records and wants to see students from Bangalore or Chennai. `isin()` can filter those cities easily.

## Business Example

A company can use `isin()` to select customers from specific cities, products from selected categories, or orders with particular statuses.

## AI/ML Example

In AI/ML, `isin()` can be useful for selecting specific categories or creating a subset of records for analysis and preprocessing.

## Python Implementation

The example below selects students who belong to Bangalore or Chennai.

## Output Explanation

Only the rows where the city matches one of the given cities are returned.

## Best Practices

- Use `isin()` when checking against multiple possible values.
- Keep the list of values clear and relevant.
- Check the filtered result before further processing.

## Common Mistakes

- Using incorrect spelling or capitalization in the values.
- Forgetting that matching is normally exact.
- Using `isin()` when a simple comparison would be enough.

## Advantages

- Makes multiple-value filtering simple.
- Easy to read and maintain.
- Useful for categorical data.

## Limitations

- It performs exact matching.
- A very large list of values can make the condition harder to manage.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "City": ["Bangalore", "Chennai", "Mumbai", "Bangalore"]
})

result = df[df["City"].isin(["Bangalore", "Chennai"])]

print(result)
```

# 7. between()

## Concept Explanation

The `between()` method is used to filter values that fall within a specific range.

It returns `True` for values between the given lower and upper limits.

By default, the boundary values are included.

## Real-world Example

Suppose Akash wants to find students whose marks are between 70 and 90. `between()` provides a simple way to filter those records.

## Business Example

A company can use `between()` to find customers whose purchase amount falls within a particular range.

## AI/ML Example

In AI/ML, `between()` can be used to select observations within a specific feature range during data exploration or preprocessing.

## Python Implementation

The example below selects students whose marks are between 80 and 90.

## Output Explanation

Only the students whose marks fall within the specified range are displayed.

## Best Practices

- Clearly define the lower and upper limits.
- Check whether the boundary values should be included.
- Use `between()` when working with simple numerical or comparable ranges.

## Common Mistakes

- Assuming the boundaries are excluded by default.
- Using the wrong column for filtering.
- Choosing a range that does not match the analysis requirement.

## Advantages

- Simple and readable range filtering.
- Reduces the need for multiple comparison operators.
- Useful for numerical and date ranges.

## Limitations

- Best suited to straightforward range conditions.
- More complex conditions may require Boolean indexing.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "Marks": [85, 72, 92, 88]
})

result = df[df["Marks"].between(80, 90)]

print(result)
```

# 8. Sorting Data

## Concept Explanation

Sorting data means arranging the rows of a DataFrame based on the values in one or more columns.

Pandas provides the `sort_values()` method for sorting data in ascending or descending order.

Sorting is useful when we want to quickly identify the highest, lowest, earliest, or latest values in a dataset.

## Real-world Example

Suppose Akash has a student dataset and wants to arrange students from the highest marks to the lowest marks. Sorting can be used to organize the records in that order.

## Business Example

A company can sort sales records by purchase amount to identify the highest-value transactions or sort products by sales performance.

## AI/ML Example

In AI/ML data exploration, sorting can help us understand the highest and lowest values of a feature and identify unusual records.

## Python Implementation

The example below sorts students based on their marks from highest to lowest.

## Output Explanation

The rows are arranged in descending order according to the `Marks` column.

## Best Practices

- Choose the correct column for sorting.
- Use ascending or descending order based on the requirement.
- Keep the original DataFrame unchanged when needed by storing the sorted result separately.

## Common Mistakes

- Confusing ascending and descending order.
- Sorting by the wrong column.
- Forgetting that `sort_values()` returns a sorted DataFrame unless `inplace=True` is used.

## Advantages

- Makes data easier to analyze.
- Helps identify highest and lowest values quickly.
- Supports sorting by multiple columns.

## Limitations

- Sorting very large datasets can require additional processing time.
- Sorting alone does not provide any statistical insight into the data.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "Marks": [85, 78, 92, 88]
})

sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)
```

# Interview Questions

### 1. What is the difference between `loc` and `iloc`?

`loc` selects data using row and column labels, while `iloc` selects data using integer positions.

### 2. What is Boolean indexing?

Boolean indexing filters a DataFrame using conditions that return `True` or `False`. Only the rows where the condition is `True` are selected.

### 3. What is the use of `query()`?

`query()` is used to filter DataFrame rows using a condition written as a string. It can make complex filtering conditions easier to read.

### 4. What is the use of `isin()`?

`isin()` checks whether values belong to a specified list of values. It is useful when filtering rows based on multiple possible values.

### 5. What does `between()` do?

`between()` filters values that fall within a specified range. By default, the lower and upper limits are included.

### 6. How do you sort a DataFrame in Pandas?

We can use `sort_values()` to sort a DataFrame based on one or more columns.

### 7. What is the difference between ascending and descending sorting?

Ascending sorting arranges values from smaller to larger, while descending sorting arranges values from larger to smaller.

### 8. Why is data selection important in AI/ML?

Data selection helps us work with only the rows and columns that are relevant to a particular analysis or Machine Learning task. It is useful during data cleaning, preprocessing, and feature selection.
