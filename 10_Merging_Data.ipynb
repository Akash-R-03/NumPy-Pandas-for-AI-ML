# 1. concat()

## Concept Explanation

The `concat()` function is used to combine two or more Pandas DataFrames along rows or columns.

It is useful when datasets have a similar structure and we want to combine them into a single DataFrame.

For example, we can combine data collected from different months, branches, or files.

## Real-world Example

Suppose Akash has student records stored in two separate DataFrames. He can use `concat()` to combine both sets of records into one DataFrame.

## Business Example

A company may have separate sales data for different months. `concat()` can combine those monthly datasets into one complete sales dataset.

## AI/ML Example

In AI/ML projects, training or analysis data may come from multiple files with the same columns. `concat()` can combine them before further preprocessing.

## Python Implementation

The example below combines two student DataFrames row by row.

## Output Explanation

The records from both DataFrames are combined into a single DataFrame.

## Best Practices

- Make sure the DataFrames have compatible columns.
- Use `ignore_index=True` when a new continuous index is required.
- Check the combined dataset after concatenation.

## Common Mistakes

- Combining DataFrames with different column structures without checking them.
- Forgetting about duplicate indexes.
- Using `concat()` when a key-based `merge()` is actually required.

## Advantages

- Simple way to combine datasets.
- Supports row-wise and column-wise concatenation.
- Useful for combining similar datasets.

## Limitations

- Different column structures can create missing values.
- It does not perform key-based matching like `merge()`.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df2 = pd.DataFrame({
    "Name": ["Kiran", "Anu"],
    "Marks": [92, 88]
})

result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

# 2. merge()

## Concept Explanation

The `merge()` function is used to combine two DataFrames using a common column or key.

It is similar to joining tables in SQL. Instead of simply placing rows together, `merge()` matches related records based on the selected column.

## Real-world Example

Suppose Akash has one DataFrame containing student names and another containing their marks. Both have a common `ID`, so they can be combined using `merge()`.

## Business Example

A company may have separate customer and order tables. A common customer ID can be used to combine the two datasets.

## AI/ML Example

In AI/ML projects, information from different datasets can be combined using a common ID before data analysis and preprocessing.

## Python Implementation

The example below combines two DataFrames using the `ID` column.

## Output Explanation

The matching records from both DataFrames are combined into one DataFrame.

## Best Practices

- Use a reliable common key for merging.
- Check for duplicate keys before merging.
- Choose the correct merge type based on the requirement.

## Common Mistakes

- Using the wrong column as the key.
- Ignoring duplicate keys.
- Choosing the wrong merge type.

## Advantages

- Combines related datasets easily.
- Similar to SQL table joins.
- Supports different types of joins.

## Limitations

- Incorrect keys can produce incorrect results.
- Duplicate keys may increase the number of rows unexpectedly.

```python
import pandas as pd

df1 = pd.DataFrame({"ID": [1, 2], "Name": ["Akash R", "Rahul"]})
df2 = pd.DataFrame({"ID": [1, 2], "Marks": [85, 78]})

result = pd.merge(df1, df2, on="ID")

print(result)
```

# 3. join()

## Concept Explanation

The `join()` method is used to combine two DataFrames based on their index.

It is useful when related information is stored in separate DataFrames but both DataFrames use the same index.

## Real-world Example

Suppose Akash has one DataFrame containing student names and another containing their marks. If both use the same student ID as the index, `join()` can combine them.

## Business Example

A company can use `join()` to combine related datasets that share the same index, such as customer details and customer activity.

## AI/ML Example

In AI/ML, `join()` can combine different sets of features that belong to the same records.

## Python Implementation

The example below joins two DataFrames using their index.

## Output Explanation

The columns from both DataFrames are combined based on their matching index.

## Best Practices

- Make sure the indexes represent the same type of records.
- Check the index before joining.
- Use `merge()` when the matching column is not the index.

## Common Mistakes

- Joining DataFrames with unrelated indexes.
- Confusing `join()` with `merge()`.
- Forgetting that `join()` is primarily index-based.

## Advantages

- Simple way to combine index-based data.
- Useful for combining related DataFrames.
- Easy to read and use.

## Limitations

- Mainly designed for index-based joining.
- Incorrect indexes can produce incorrect results.

```python
import pandas as pd

df1 = pd.DataFrame({"Name": ["Akash R", "Rahul"]}, index=[1, 2])
df2 = pd.DataFrame({"Marks": [85, 78]}, index=[1, 2])

result = df1.join(df2)

print(result)
```

# 4. append()

## Concept Explanation

The `append()` method was previously used in Pandas to add rows from one DataFrame to another.

However, `DataFrame.append()` was deprecated in Pandas 1.4 and removed in Pandas 2.0.

The recommended alternative is `pd.concat()`.

So, in modern Pandas projects, we should use `concat()` instead of `append()`.

## Real-world Example

Suppose Akash wants to add a new student record to an existing DataFrame. Instead of using the old `append()` method, he can create the new record and combine it using `concat()`.

## Business Example

A company may receive new sales records and need to add them to an existing dataset. `concat()` can be used to combine the records.

## AI/ML Example

In AI/ML workflows, new records may need to be added to an existing dataset. `concat()` provides the modern way to combine them.

## Python Implementation

The example below shows the recommended alternative to the old `append()` method.

## Output Explanation

The new student record is added to the existing DataFrame using `concat()`.

## Best Practices

- Use `pd.concat()` instead of `append()`.
- Avoid old Pandas code that depends on `DataFrame.append()`.
- Check the Pandas version when working with older notebooks or projects.

## Common Mistakes

- Trying to use `df.append()` in modern Pandas versions.
- Using `append()` when `concat()` is the recommended option.
- Forgetting to use `ignore_index=True` when a continuous index is required.

## Advantages

- `concat()` is the supported modern approach.
- Works with multiple DataFrames.
- Easy to understand and maintain.

## Limitations

- Older tutorials may still show `append()`, which can cause errors in modern Pandas.

```python
import pandas as pd

df = pd.DataFrame({"Name": ["Akash R"], "Marks": [85]})
new_row = pd.DataFrame({"Name": ["Rahul"], "Marks": [78]})

result = pd.concat([df, new_row], ignore_index=True)

print(result)
```

# 5. combine_first()

## Concept Explanation

The `combine_first()` method is used to fill missing values in one DataFrame or Series using values from another.

If a value already exists in the first DataFrame, it is kept. If it is missing, Pandas uses the corresponding value from the second DataFrame.

## Real-world Example

Suppose Akash has student information from two sources. If some marks are missing in the first dataset but available in the second dataset, `combine_first()` can fill those missing values.

## Business Example

A company may have customer information from two different sources. `combine_first()` can help fill missing information from one source using available information from another.

## AI/ML Example

In AI/ML data preparation, `combine_first()` can help fill missing values when a reliable alternative dataset contains the required information.

## Python Implementation

The example below fills a missing mark using the value from another DataFrame.

## Output Explanation

The existing value is kept, while the missing value is filled from the second DataFrame.

## Best Practices

- Make sure both DataFrames have matching indexes.
- Use a reliable source for filling missing values.
- Check the result after combining.

## Common Mistakes

- Using DataFrames with unrelated indexes.
- Assuming existing values will be replaced.
- Filling missing values without checking whether the second source is reliable.

## Advantages

- Simple way to fill missing values from another dataset.
- Preserves existing non-missing values.
- Useful when combining information from multiple sources.

## Limitations

- It works based on matching indexes.
- It does not automatically determine whether the replacement value is correct.

```python
import pandas as pd

df1 = pd.DataFrame({"Marks": [85, None]}, index=[1, 2])
df2 = pd.DataFrame({"Marks": [90, 78]}, index=[1, 2])

result = df1.combine_first(df2)

print(result)
```
