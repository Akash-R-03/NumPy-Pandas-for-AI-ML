# 1. What is Pandas?

## Concept Explanation

# 3. Series

## Real-world Example

Suppose Akash has the marks of five subjects. These marks can be stored as a Pandas Series, where each value represents the mark for one subject.

## Business Example

## Best Practices

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df.to_csv("students.csv", index=False)

data = pd.read_csv("students.csv")

print(data)
```

# 10. Writing Data

## Concept Explanation

Writing data means saving a Pandas DataFrame into a file so that it can be used later.

Pandas provides functions such as `to_csv()` to save DataFrames into CSV files.

This is useful when we have processed or analyzed data and want to store the result.

## Real-world Example

Suppose Akash has cleaned and organized some student information. He can save the DataFrame as a CSV file and use it later without creating the data again.

## Business Example

A company can save processed sales or customer data into a CSV file and share it with other teams or use it for reporting.

## AI/ML Example

After cleaning and preprocessing a dataset, it can be saved so that the prepared data can be reused for further analysis or Machine Learning work.

## Python Implementation

The code below creates a DataFrame and saves it as a CSV file.

## Output Explanation

The DataFrame is saved as `students.csv` in the current working directory.

## Best Practices

- Use `index=False` when the DataFrame index is not required in the CSV file.
- Use clear and meaningful file names.
- Check that the file was saved correctly.

## Common Mistakes

- Saving the file to an unexpected location.
- Accidentally saving the DataFrame index as an extra column.
- Overwriting an important file without checking its name.

## Advantages

- Easy way to save processed data.
- CSV files are simple to share and use.
- Saved data can be loaded again using Pandas.

## Limitations

- CSV files do not preserve all Pandas-specific information.
- Large CSV files can take more storage and processing time compared with some specialized formats.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, 78]
})

df.to_csv("students.csv", index=False)

print("Data saved successfully.")
```
