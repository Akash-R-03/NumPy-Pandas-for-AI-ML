# 1. isnull()

## Concept Explanation

The `isnull()` method is used to identify missing or null values in a Pandas Series or DataFrame.

It returns `True` where a value is missing and `False` where a value is available.

It is one of the first methods we can use when checking the quality of a dataset.

## Real-world Example

Suppose Akash has a student dataset where some students have not provided their marks. `isnull()` can help identify those missing values.

## Business Example

A company can use `isnull()` to find missing customer information such as phone numbers, email addresses, or order details.

## AI/ML Example

In AI/ML, identifying missing values is an important step before training a model because many algorithms cannot directly work with missing data.

## Python Implementation

The example below checks for missing values in a small DataFrame.

## Output Explanation

`True` indicates that the value is missing, while `False` indicates that the value is available.

## Best Practices

- Check missing values before performing analysis.
- Inspect missing values column by column.
- Understand why values are missing before deciding how to handle them.

## Common Mistakes

- Treating every missing value as an error.
- Removing missing values without understanding the reason.
- Checking only one column when the entire dataset needs inspection.

## Advantages

- Simple way to identify missing values.
- Useful during data cleaning.
- Works with both Series and DataFrames.

## Limitations

- It only identifies missing values; it does not fix them.
- Further methods are required to handle the missing data.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, None]
})

print(df.isnull())
```

# 2. notnull()

## Concept Explanation

The `notnull()` method is used to identify values that are available or not missing in a Pandas Series or DataFrame.

It returns `True` where a value is present and `False` where a value is missing.

It is the opposite of `isnull()`.

## Real-world Example

Suppose Akash has a student dataset and wants to identify which students have provided their marks. `notnull()` can be used to find the available values.

## Business Example

A company can use `notnull()` to identify customers who have provided important information such as email addresses or phone numbers.

## AI/ML Example

In AI/ML, `notnull()` can help identify records that contain the required values before performing data preprocessing or analysis.

## Python Implementation

## Output Explanation

`True` indicates that a value is present, while `False` indicates that the value is missing.

## Best Practices

- Use `notnull()` when you specifically want to identify available values.
- Check important columns before analysis.
- Combine it with filtering when necessary.

## Common Mistakes

- Confusing `notnull()` with `isnull()`.
- Assuming `True` means the value is missing.
- Using it without checking which columns are relevant.

## Advantages

- Simple way to identify available values.
- Useful for filtering complete records.
- Works with Series and DataFrames.

## Limitations

- It only identifies available values.
- It does not fix or fill missing values.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, None]
})

print(df.notnull())
```

# 3. dropna()

## Concept Explanation

The `dropna()` method is used to remove rows or columns that contain missing values from a Pandas DataFrame.

It is useful when missing values cannot be filled reliably or when the affected records are not important for the analysis.

## Real-world Example

Suppose Akash has a student dataset where one student's marks are missing. If that record is not required, `dropna()` can be used to remove the row.

## Business Example

A company may remove records with missing critical information when those records cannot be used for a particular analysis.

## AI/ML Example

In AI/ML preprocessing, rows with missing values may sometimes be removed when the number of missing records is small and removing them will not significantly affect the dataset.

## Output Explanation

The row containing the missing value is removed from the DataFrame.

## Best Practices

- Check the amount of missing data before removing rows.
- Use `dropna()` when removing the affected records is reasonable.
- Consider the impact of data loss before using it.

## Common Mistakes

- Removing too many rows.
- Using `dropna()` without understanding why values are missing.
- Losing important information unnecessarily.

## Advantages

- Simple way to remove missing data.
- Helps create a complete dataset.
- Easy to apply.

## Limitations

- Can result in data loss.
- Not suitable when many important records contain missing values.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, None]
})

result = df.dropna()

print(result)
```

# 4. fillna()

## Concept Explanation

The `fillna()` method is used to replace missing values with a specified value.

It is useful when we want to keep the records but need to provide a suitable value for the missing data.

## Real-world Example

Suppose Akash has a student dataset where one student's marks are missing. If the average marks are 80, the missing value can be filled with 80.

## Business Example

A company can use `fillna()` to replace missing numerical values with an appropriate value such as the average, median, or a predefined value.

## AI/ML Example

In AI/ML preprocessing, missing numerical values can be filled using methods such as mean or median imputation instead of removing the entire record.

## Python Implementation

The example below replaces the missing marks value with `80`.

## Output Explanation

The missing value is replaced with the specified value while the other values remain unchanged.

## Best Practices

- Choose the replacement value based on the meaning of the data.
- Use mean or median carefully when appropriate.
- Check the data after filling missing values.

## Common Mistakes

- Filling all missing values with an arbitrary number.
- Using the mean when the data contains strong outliers.
- Forgetting to verify the filled values.

## Advantages

- Prevents unnecessary loss of records.
- Simple way to handle missing values.
- Supports different replacement strategies.

## Limitations

- A poorly chosen replacement value can introduce bias.
- Filling values does not recover the original information.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul"],
    "Marks": [85, None]
})

result = df.fillna({"Marks": 80})

print(result)
```

# 5. interpolate()

## Concept Explanation

The `interpolate()` method is used to estimate and fill missing values based on the values around them.

For numerical data, Pandas can calculate a value between known values instead of simply replacing the missing value with a fixed number.

## Real-world Example

Suppose Akash has daily temperature readings where one day's value is missing. If the values before and after it are available, interpolation can estimate the missing value.

## Business Example

A company can use interpolation to estimate missing numerical measurements, such as sales, production, or sensor readings, when values follow a continuous pattern.

## AI/ML Example

In AI/ML, interpolation can be useful for filling missing values in time-series or continuous numerical data before further analysis or modeling.

## Python Implementation

The example below fills a missing value between 80 and 90.

## Output Explanation

The missing value is estimated using the surrounding values.

## Best Practices

- Use interpolation when the data has a meaningful order or continuous pattern.
- Understand the nature of the data before applying it.
- Check the interpolated values after filling them.

## Common Mistakes

- Using interpolation for categorical data.
- Assuming the estimated value is always the actual value.
- Using interpolation when the data has no meaningful order.

## Advantages

- Provides a more meaningful estimate than a fixed value in suitable datasets.
- Useful for continuous numerical data.
- Helpful for time-series data.

## Limitations

- The estimated value may not represent the actual value.
- It is not suitable for every type of dataset.
- Poorly chosen interpolation methods can produce misleading results.

```python
import pandas as pd

df = pd.DataFrame({"Marks": [80, None, 90]})

result = df.interpolate()

print(result)
```

# 6. Missing Value Handling Strategies

## Concept Explanation

Missing Value Handling means deciding what to do when a dataset contains missing or null values.

There is no single method that works for every dataset. Depending on the situation, we can remove missing records, fill them with suitable values, or estimate them using methods such as interpolation.

Common strategies include:

- **Remove:** Use `dropna()` when missing records are not important.
- **Fill:** Use `fillna()` with a suitable value such as mean, median, or a fixed value.
- **Interpolate:** Use `interpolate()` when missing numerical values can be estimated from surrounding values.
- **Keep:** Sometimes missing values are meaningful and should be left unchanged.

## Real-world Example

Suppose Akash has student data where some marks are missing. He can decide whether to remove those records, fill the missing marks, or estimate them depending on the situation.

## Business Example

A company may have missing customer or sales information. The appropriate strategy depends on how important the missing information is and why it is missing.

## AI/ML Example

In AI/ML, missing values should be handled before training many models. The chosen strategy can affect the quality of the data and the model's performance.

## Python Implementation

The example below replaces a missing marks value with the median of the available marks.

## Output Explanation

The missing value is replaced with the median value calculated from the available marks.

## Best Practices

- Understand why the values are missing before handling them.
- Choose a strategy based on the type and meaning of the data.
- Avoid unnecessary removal of records.
- Check the dataset after handling missing values.

## Common Mistakes

- Filling every missing value with zero without considering its meaning.
- Removing a large number of records unnecessarily.
- Using interpolation for categorical data.
- Calculating replacement values using inappropriate data.

## Advantages

- Helps create cleaner datasets.
- Makes data suitable for further analysis.
- Can prevent problems during Machine Learning.

## Limitations

- Some methods may introduce bias.
- Filled or estimated values may not represent the actual values.
- Removing records can reduce the amount of available data.

```python
import pandas as pd

df = pd.DataFrame({"Marks": [80, 90, None]})

df["Marks"] = df["Marks"].fillna(df["Marks"].median())

print(df)
```
