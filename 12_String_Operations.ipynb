# 6. replace()

## Concept Explanation

The `replace()` method is used to replace one part of a string with another value.

It is useful for correcting, cleaning, or standardizing text.

## Real-world Example

Suppose Akash has a text value containing `B'lore` and wants to replace it with `Bangalore`.

## Business Example

A company can use `replace()` to correct inconsistent product names, city names, categories, or other text values.

## AI/ML Example

In Natural Language Processing (NLP), `replace()` can be used during text cleaning to remove or replace unwanted characters or words.

## Python Implementation

The example below replaces `B'lore` with `Bangalore`.

## Output Explanation

The specified text is replaced with the new value.

## Best Practices

- Clearly identify the text that needs to be replaced.
- Check the result after replacement.
- Keep the original data when it may be needed later.

## Common Mistakes

- Assuming `replace()` changes the original string.

## Advantages

- Simple text-cleaning method.
- Useful for correcting inconsistent values.

# 7. Regular Expressions (Regex)

## Concept Explanation

Regular Expressions, commonly called Regex, are patterns used to search, match, and extract specific text from strings.

Regex is useful when the text follows a particular pattern, such as email addresses, phone numbers, dates, or special characters.

Python provides the `re` module for working with Regular Expressions.

## Real-world Example

Suppose Akash has a text containing an email address and wants to check whether an email pattern exists. Regex can be used to find it.

## Business Example

A company can use Regex to validate email addresses, phone numbers, employee IDs, or other structured text entered by users.

## AI/ML Example

In Natural Language Processing (NLP), Regex can be used to clean text, extract specific patterns, or remove unwanted characters before further processing.

## Python Implementation

The example below uses Regex to find an email address in a text.

## Output Explanation

The matching email address is extracted from the given text.

## Best Practices

- Keep Regex patterns simple when possible.
- Test patterns with different types of input.
- Use Regex only when pattern-based text processing is required.

## Common Mistakes

- Writing overly complex patterns.
- Forgetting special characters have specific meanings in Regex.
- Not testing the pattern against different inputs.

## Advantages

- Powerful for pattern-based text processing.
- Useful for searching and extracting information.
- Helpful for data cleaning and validation.

## Limitations

- Complex Regex patterns can be difficult to read.
- Incorrect patterns may produce unexpected results.
- Regex is not always the best choice for complex text processing.

```python
import re

text = "Akash's email is akash@example.com"

result = re.findall(r"\S+@\S+", text)

print(result)
```
