# 13. Copy vs View

## Concept Explanation

In NumPy, a copy and a view may look similar, but they work differently.

A **copy** creates a separate array. If we change the copy, the original array is not affected.

A **view** shares the same underlying data with the original array. So, changing the view can also change the original array.

## Real-world Example

Think of a copy like making a photocopy of a document. You can change the photocopy without changing the original.

A view is more like looking at the same document from another place. Changes to the shared document can be seen from both places.

## Business Example

When working with a large business dataset, creating unnecessary copies can use more memory. A view can sometimes be useful when we only need another way to access the same data.

## AI/ML Example

In AI/ML data preprocessing, understanding copy and view is important when modifying arrays. Accidentally changing the original data can affect later steps in the workflow.

## Python Implementation

The example below shows the difference between a copy and a view.

## Output Explanation

Changing the copy does not change the original array. However, changing the view also changes the original array because they share the same data.

## Best Practice

Use `.copy()` when you need an independent array that can be changed safely.

## Common Mistake

Assuming that every new array variable contains completely separate data.

## Advantage

Views can avoid unnecessary copying and can save memory.

## Limitation

Views can cause unexpected changes to the original data if we are not careful.

```python
import numpy as np

original = np.array([10, 20, 30])

# Copy
copied = original.copy()
copied[0] = 100

# View
view = original.view()
view[1] = 200

print("Original:", original)
print("Copy:", copied)
print("View:", view)
```
