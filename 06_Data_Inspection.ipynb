# 1. head()

## Concept Explanation

The `head()` method is used to view the first few rows of a Pandas DataFrame.

By default, `head()` displays the first 5 rows. We can also provide a number to display a specific number of rows.

It is one of the first methods we can use when inspecting a new dataset.

## Real-world Example

Suppose Akash receives a large customer dataset. Instead of displaying the entire dataset, he can use `head()` to quickly see how the first few records look.

## Business Example

A company can use `head()` to quickly check the first few sales or customer records before starting an analysis.

## AI/ML Example

In an AI/ML project, `head()` is useful for getting a quick look at a newly loaded dataset and checking whether the columns and values appear as expected.

## Python Implementation

The example below creates a small DataFrame and displays its first three rows.

## Output Explanation

The output shows only the first three rows of the DataFrame.

## Best Practices

- Use `head()` when initially exploring a dataset.
- Specify the number of rows when you need a particular amount of data.
- Use it along with other inspection methods for a better understanding of the dataset.

## Common Mistakes

- Assuming `head()` shows the complete dataset.
- Forgetting that the default number of rows is 5.

## Advantages

- Quick way to inspect data.
- Useful for large datasets.
- Simple and easy to understand.

## Limitations

- Only shows a small portion of the dataset.
- It may not reveal issues that occur later in the dataset.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu"],
    "Marks": [85, 78, 92, 88]
})

print(df.head(3))
```

# 3. sample()

## Concept Explanation

The `sample()` method is used to randomly select rows from a Pandas DataFrame.

Unlike `head()` and `tail()`, which always show rows from a fixed position, `sample()` selects rows randomly.

By default, it selects one row. We can specify how many rows we want to select.

## Real-world Example

Suppose Akash has thousands of customer records and wants to quickly look at a few random customers. `sample()` can select those records without checking the complete dataset.

## Business Example

A company can randomly select customers for surveys, quality checks, or small data reviews.

## AI/ML Example

In AI/ML, random samples can be useful for exploring a dataset and checking whether randomly selected records look reasonable.

## Python Implementation

The example below randomly selects two rows from the DataFrame.

## Output Explanation

The output contains two randomly selected rows from the dataset.

## Best Practices

- Use `random_state` when you need reproducible results.
- Use `sample()` when you want a random view of the data.
- Choose an appropriate sample size.

## Common Mistakes

- Assuming `sample()` always returns the same rows.
- Forgetting to use `random_state` when reproducibility is required.

## Advantages

- Provides a quick random view of the data.
- Useful for large datasets.
- Helpful for sampling and testing.

## Limitations

- A small sample may not represent the complete dataset.
- Results can change between runs without a fixed random state.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu", "Ravi"],
    "Marks": [85, 78, 92, 88, 81]
})

print(df.sample(2, random_state=42))
```

# 11. value_counts()

## Concept Explanation

The `value_counts()` method is used to count how many times each unique value appears in a Pandas Series.

It is especially useful for understanding the frequency of categories or repeated values in a column.

## Real-world Example

Suppose Akash has student data containing the city of each student. `value_counts()` can show how many students belong to each city.

## Business Example

A company can use `value_counts()` to find how many customers belong to each location, product category, or order status.

## AI/ML Example

In AI/ML, `value_counts()` can help understand the distribution of categorical target values or features. This can also help identify whether some categories are much more common than others.

## Python Implementation

The example below counts how many students belong to each city.

## Output Explanation

The output shows each unique city and the number of times it appears.

## Best Practices

- Use `value_counts()` to inspect categorical data.
- Check category frequencies before preprocessing.
- Look for highly imbalanced categories when working with Machine Learning data.

## Common Mistakes

- Confusing `value_counts()` with `nunique()`.
- Forgetting that `nunique()` gives the number of unique values, while `value_counts()` gives the frequency of each value.

## Advantages

- Quickly shows the frequency of values.
- Useful for categorical data analysis.
- Helps identify common and rare categories.

## Limitations

- It only summarizes the selected Series.
- Frequency counts alone do not explain why some values occur more often than others.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akash R", "Rahul", "Kiran", "Anu", "Ravi"],
    "City": ["Bangalore", "Chennai", "Bangalore", "Chennai", "Bangalore"]
})

print(df["City"].value_counts())
```

# Interview Questions

### 1. What is the purpose of `head()` and `tail()`?

### 6. What is the difference between `nunique()` and `value_counts()`?

### 7. What is the purpose of `dtypes`?

`dtypes` shows the data type of each column in a DataFrame. It is useful when checking and preparing data for analysis or Machine Learning.
