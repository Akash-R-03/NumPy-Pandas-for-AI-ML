# 3. Date Parsing

## Concept Explanation

Date parsing means converting a date stored as text into a proper date or datetime value that Python or Pandas can understand and work with.

This is useful because dates are often stored in different formats in real-world datasets.

## Real-world Example

Suppose Akash receives a dataset where dates are stored as text such as `20-08-2026`. He can parse these values into proper datetime values before performing date-based operations.

## Business Example

A company may receive dates in different formats from different systems. Date parsing helps convert them into a consistent format for reporting and analysis.

## AI/ML Example

In AI/ML projects, date parsing is important when working with time-based datasets because date values may need to be converted before extracting features or analyzing trends.

## Python Implementation

The example below converts `20-08-2026` from text into a Pandas datetime value.

## Output Explanation

The text date is converted into a proper datetime representation.

## Best Practices

- Check the format of the original date before parsing.
- Keep date formats consistent.
- Handle invalid or missing dates carefully.

## Common Mistakes

- Using the wrong date format.
- Assuming all dates follow the same format.
- Ignoring invalid date values.

## Advantages

- Converts text dates into usable datetime values.
- Makes date calculations easier.
- Helps standardize date data.

## Limitations

- Incorrect formats can cause parsing errors.
- Different date formats may require different parsing options.

```python
import pandas as pd

date = pd.to_datetime("20-08-2026", dayfirst=True)

print(date)
```

# 4. Timedelta

## Concept Explanation

`Timedelta` represents a duration or the difference between two dates or times.

It is useful when we want to add or subtract a specific amount of time from a date.

## Real-world Example

Suppose Akash wants to calculate a future date after a particular number of days. `Timedelta` can be used to add that duration to the original date.

## Business Example

A company can use `Timedelta` to calculate delivery dates, subscription periods, or payment deadlines.

## AI/ML Example

In AI/ML, `Timedelta` can be useful for calculating the time between events, such as the time between transactions or customer activities.

## Python Implementation

The example below adds 5 days to 20 August 2026.

## Output Explanation

The output shows the new date after adding the specified duration.

## Best Practices

- Use `Timedelta` for date and time differences.
- Choose the correct unit, such as days, hours, or minutes.
- Check the resulting date after performing calculations.

## Common Mistakes

- Confusing a date with a duration.
- Using the wrong time unit.
- Performing calculations on text instead of datetime values.

## Advantages

- Makes date calculations simple.
- Supports different time units.
- Useful for calculating durations and future dates.

## Limitations

- It represents a duration, not a specific calendar date.
- More complex calendar calculations may require additional handling.

```python
import pandas as pd

date = pd.Timestamp("2026-08-20")
new_date = date + pd.Timedelta(days=5)

print(new_date)
```

# 5. Time Zones

## Concept Explanation

A time zone represents the local time used in a particular geographical region.

Pandas allows us to create timezone-aware datetime values and convert them between different time zones.

This is useful when working with data collected from different locations around the world.

## Real-world Example

Suppose Akash is working with an event scheduled in India. The event time can be stored with the India time zone and converted to another time zone when needed.

## Business Example

A global company can use time zones to correctly manage meetings, transactions, orders, and customer activities across different countries.

## AI/ML Example

In AI/ML, time zones are important when working with global time-series data, such as user activity or transactions from different regions.

## Python Implementation

The example below creates a timestamp for 20 August 2026 at 10:30 AM in India and converts it to UTC.

## Output Explanation

The first output shows the time in India, while the second shows the same moment in UTC.

## Best Practices

- Store and process timezone-aware values when working with multiple regions.
- Use a consistent timezone for systems that collect global data.
- Convert to the required local timezone when presenting results.

## Common Mistakes

- Ignoring time zones when comparing times from different locations.
- Treating timezone-aware and timezone-naive values as the same.
- Converting a time without considering the original timezone.

## Advantages

- Helps manage dates and times across different regions.
- Useful for global applications.
- Makes time-based comparisons more accurate.

## Limitations

- Time zone conversions can add complexity.
- Incorrect timezone information can lead to incorrect results.

```python
import pandas as pd

time = pd.Timestamp("2026-08-20 10:30", tz="Asia/Kolkata")

print(time)
print(time.tz_convert("UTC"))
```
